### 2. What is a graph?

A **node** represents a canonical street-network junction or terminal. An **edge** represents a physical street connection between two nodes. A node's **degree** is the number of incident physical edges. A **path** is an ordered sequence of connected nodes and edges. A **connected component** is a maximal set of nodes that can reach one another through paths.

### 3. What is NetworkX?

NetworkX supplies general graph structures and network-science algorithms in Python. GeoGami uses a NetworkX `MultiGraph` as the scientific representation of undirected physical streets. The `Multi` prefix means that distinct canonical edges can, when scientifically valid, connect the same node pair using different edge keys.

### 4. What is OSMnx?

OSMnx provides street-network-specific geospatial functionality built around NetworkX, GeoPandas, and Shapely. It is often used to download OpenStreetMap networks, but its public `convert.graph_from_gdfs` workflow also accepts custom node and edge GeoDataFrames. GeoGami uses that custom-data workflow.

### 5. Why Env39 is not downloaded from OpenStreetMap

Env39 is a synthetic, local GeoGami environment originating in manual/research QGIS digitization. Its authoritative route is **canonical GeoPackage → GeoPandas → `graph_from_gdfs`**, not an OSM place download. Internet access is neither required nor used by these notebooks.

### 6. Scientific `MultiGraph`

The canonical scientific graph is an undirected `nx.MultiGraph`. It contains one graph edge for each physical canonical street. Current Env39 therefore has 46 nodes and 69 physical edges.

### 7. OSMnx `MultiDiGraph`

OSMnx commonly represents movement possibilities as directed arcs in an `nx.MultiDiGraph`. Because the current canonical streets are undirected, the adapter creates one forward and one reverse arc per physical street while retaining the physical canonical edge ID on both.

### 8. Why 69 physical streets become 138 directed arcs

One physical street:

```text
N001 -------- N014
```

becomes two reciprocal analysis arcs:

```text
N001 → N014
N001 ← N014
```

Thus **69 physical canonical streets = 138 directed analysis arcs**. The value 138 must never be interpreted as 138 physical streets.

### 9. Coordinate system and units

The network uses **GeoGami Local Cartesian** coordinates. These are not longitude/latitude and are not EPSG:4326. Until a scientifically justified real-world scale is established, distances are reported as **local units**, not metres.

### 10. Geographic OSMnx functions deliberately not used

`ox.distance.add_edge_lengths` is not the authoritative length source because its geographic endpoint assumptions are inappropriate here. `ox.bearing.add_edge_bearings` is also not used for this local Cartesian network. Authoritative edge length is planar `LineString.length` in local units.
