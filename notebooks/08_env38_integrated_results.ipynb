# Env38 - Integrated Scientific Results

Professor-facing synthesis of the accepted Env38 topology, geometry, and orientation artifacts.

## 1. Purpose and scientific boundary
Load the production integration package, explain its comparison-ready schema, and summarize Env38 descriptively. This notebook does not calculate new scientific metrics or compare Env38 with Env39.

## 2. Dynamic accepted-run and provenance-chain resolution
Canonical and analysis pointers are resolved and cross-checked before integration. The active run identifier is never hardcoded.

```python
from pathlib import Path
import json, sys

def find_repository(start):
    for candidate in (start, *start.parents):
        if (candidate / 'environment.yml').is_file() and (candidate / 'config' / 'environments.yaml').is_file():
            return candidate
    raise RuntimeError('Repository root not found')
REPOSITORY = find_repository(Path.cwd().resolve())
if str(REPOSITORY / 'src') not in sys.path:
    sys.path.insert(0, str(REPOSITORY / 'src'))
from geogami_morphology.environments import get_environment
from geogami_morphology.integrated import integrate_results
ENVIRONMENT, CANONICAL_RUN = 'env38', 'latest'
environment = get_environment(ENVIRONMENT, repository_root=REPOSITORY)
canonical_pointer = json.loads(environment.canonical_latest.read_text(encoding='utf-8'))
analysis_pointer = json.loads((environment.analysis_root / 'latest.json').read_text(encoding='utf-8'))
assert canonical_pointer['run_id'] == analysis_pointer['canonical_run_id']
assert (REPOSITORY / canonical_pointer['canonical_path']).resolve() == (REPOSITORY / analysis_pointer['canonical_path']).resolve()
print('ENV38_INTEGRATED_POINTER_RESOLUTION: PASS')
```

## 3. Loading accepted source artifacts
`integrate_results(..., publish=False)` verifies the canonical identity and the accepted graph, topology, geometry, configuration, and artifact hashes before loading stored summaries.

```python
result = integrate_results(ENVIRONMENT, canonical_run=CANONICAL_RUN, latest_path=environment.canonical_latest, project_root=REPOSITORY, publish=False)
assert result.selection.run_id == canonical_pointer['run_id']
assert len(result.metrics) == 39
assert len(result.core_metrics) == 19
print('ENV38_INTEGRATED_SOURCE_LOADING: PASS')
```

## 4. Metric families
The 39 comparison-ready metrics are classified as `topology_controlled`, `geometry_weighted_network`, `geometric_morphology`, or `orientation_morphology`.

## 5. Comparison roles
Roles distinguish `topology_control`, `geometry_sensitive`, `orientation_sensitive`, and `context_only` evidence. Classification prevents connectivity controls from being interpreted as direct geometry outcomes.

## 6. Core result set
Nineteen core metrics provide a concise baseline; twenty supplementary metrics preserve additional control and distribution context in the comparison-ready file.

```python
for row in result.core_metrics:
    print(f"{row['metric_id']}: {row['value']} {row['units']} [{row['comparison_role']}]")
print('ENV38_INTEGRATED_METRICS: 39 total | 19 core | PASS')
```

## 7. Topology-control interpretation
Node/edge counts, connectedness, degree composition, and cycle rank document the fixed connectivity. They are experimental controls, not performance scores.

## 8. Geometry-weighted network interpretation
Weighted shortest-path length uses GeoGami Local Cartesian coordinates, which are not longitude/latitude. It may vary with geometry even when topology remains identical. Distances are reported in local units; do not call them metres unless a justified scale transformation is established.

## 9. Geometric morphology interpretation
Total and mean length plus edge and network circuity describe the realized planar street geometry.

## 10. Orientation morphology interpretation
Chord and segment-length-weighted entropy and fourfold order describe axial directionality under the accepted 36-bin, natural-log, and weighting definitions.

## 11. Five-figure evidence catalog
The accepted catalog contains the canonical physical network, degree structure, edge-circuity distribution, chord-orientation rose, and segment-length-weighted orientation distribution.

```python
assert len(result.figure_catalog) == 5
for figure in result.figure_catalog:
    path = REPOSITORY / figure['path']
    assert path.is_file()
    print(f"{figure['figure_id']}: {figure['label']} -> {figure['path']}")
print('ENV38_INTEGRATED_FIGURES: 5 | PASS')
```

## 12. Integrated artifact inventory
The accepted directory contains `comparison_ready_metrics.csv`, `core_metrics.csv`, `core_metrics.json`, `integrated_methodology.json`, `env38_baseline_summary.md`, the integrated manifest, and its figure directory.

```python
required = ['comparison_ready_metrics.csv','core_metrics.csv','core_metrics.json','integrated_methodology.json','env38_baseline_summary.md','integrated_analysis_manifest.json']
assert all((result.output_dir / name).is_file() for name in required)
print('ENV38_INTEGRATED_ARTIFACTS: ' + ', '.join(required))
```

## 14. Limitations and responsible interpretation
This baseline is descriptive. Higher or lower values are not intrinsically better or worse, and they do not establish causal effects on navigation, behavior, or experience. Results inherit local-coordinate, graph, segmentation, binning, and weighting assumptions.
